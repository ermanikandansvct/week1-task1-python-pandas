```python
students = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Name": ["Arun", "Bala", "Chandru", "Deepak"]
})

marks = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4],
    "Marks": [85, 78, 92, 74]
})

print("Students Data:")
display(students)

print("Marks Data:")
display(marks)
```

```python
merged_df = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="inner"
)

merged_df
```

```python
left_merged = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="left"
)

left_merged
```

```python
right_merged = pd.merge(
    students,
    marks,
    on="Student_ID",
    how="right"
)

right_merged
```

```python
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar"],
    "Product_A": [100, 120, 140],
    "Product_B": [80, 90, 110]
})

sales
```

```python
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar"],
    "Product_A": [100, 120, 140],
    "Product_B": [80, 90, 110]
})

sales
```

```python
# Create sales data
sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar"],
    "Product_A": [100, 120, 140],
    "Product_B": [80, 90, 110]
})

# Reshape using melt
long_data = pd.melt(
    sales,
    id_vars=["Month"],
    var_name="Product",
    value_name="Sales"
)

# Reshape back using pivot
pivot_data = long_data.pivot(
    index="Month",
    columns="Product",
    values="Sales"
)

pivot_data
```
